```python
'''
Python的代码块不使⽤⼤括号（{}）来控制类，函数以及其他逻辑判断，
其最具特⾊的就是⽤缩进来写模块。
缩进的空⽩数量是可变的（建议⽤4个空格）,
但是所有代码块语句必须包含相同的缩进空⽩数量，这个必须严格执⾏。
'''
def squares(n=10):
    print('Generating squares from 1 to %d'% (n**2))
    #这是一个循环代码块
    for i in range(1,n+1):
        print(i**2,end=' ')
        #循环代码块结束
    #函数代码块结束
gen=squares()
```

```python
dt.date()
```

```python
dt.time()
```

```python
dt.day
```

```python
dt.month
```

```python
# ⽇期和时间可以进⾏运算差值：
dt1=datetime(2015,8,20,10,18)
dt2=datetime(2014,8,20,0,18,22)
dt1-dt2
```

```python
#可⽤strftime⽅法将datetime格式化为字符串：
dt2.strftime('%m/%d/%Y %H:%M')
```

```python
#字符串可⽤strptime函数转换为datetime对象：
datetime.strptime('20150820','%Y%m%d')
```

```python
#替换datetime中的部分字段：
dt1.replace(year=2016,hour=12)
```
